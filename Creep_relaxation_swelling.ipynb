# Creep
Creep is the slow, time-dependent, and permanent deformation of a material when subjected to constant stress over a long period, especially at high temperatures. This is called "forward creep".

Primary creep: Material resists deformation — it "fights back," so the creep rate slows down.

Secondary creep: Material reaches a steady rhythm — deformation happens at a constant rate.

Reverse creep or recovery = Deformation reducing if the load is removed or reversed.

# Stress relaxation
Stress relaxation is the decrease of stress over time when a material is held at constant strain.

➔ Imagine you stretch a material to a fixed length and then hold it there.

➔ Over time, the material relaxes and the internal stress drops, even though you are not allowing it to contract.


# Swelling	
Permanent, internal, volume expansion due to environmental or internal effects (absorption, radiation, chemistry), independent of external loading.

➔ Swelling can combine with creep and plasticity — for example, swelling can cause internal stresses that then lead to plastic deformation or accelerated creep.


# plasticity-Creep model


$$
\dot{\boldsymbol{y}}=f(\boldsymbol{y},\dot{\boldsymbol{\varepsilon}},T,\dot{T},t)\dot{\boldsymbol{x}}\\
\left\{
   \begin{aligned}
    \boldsymbol{y} &= [\boldsymbol{\sigma}^{pl}, \boldsymbol{h} ]\\
    \boldsymbol{x} &= [\boldsymbol{\varepsilon}, T, t]
\end{aligned}
    \right.
$$

$$
\dot{\boldsymbol{\varepsilon}}^{cr} = g(\boldsymbol{\sigma}^{pl},\boldsymbol{\varepsilon}^{cr},T)\dot{t}
$$

$$
\boldsymbol{\varepsilon} = \boldsymbol{\varepsilon} + \boldsymbol{\varepsilon}^{cr}
$$
The above nonlinear equations are solved simultaneously to capture both history-dependent plasticity and time-dependent creep behavior of material. $f$ and $g$ are both neural operators learned from experimental/synthetic data. 



# plasticity-stress relaxation model


$$
\dot{\boldsymbol{y}}=f(\boldsymbol{y},\dot{\boldsymbol{\varepsilon}},T,\dot{T},t)\dot{\boldsymbol{x}}\\
\left\{
   \begin{aligned}
    \boldsymbol{y} &= [\boldsymbol{\sigma}^{pl}, \boldsymbol{h} ]\\

\end{aligned}
    \right.
$$

$$
\dot{\boldsymbol{\sigma}}^{re} = g(\boldsymbol{\sigma}^{pl},\boldsymbol{\varepsilon},T)\dot{t}
$$

$$
\boldsymbol{\sigma}^{pl} = \boldsymbol{\sigma}^{pl} + \boldsymbol{\sigma}^{re}
$$
The above nonlinear equations are solved simultaneously to capture both history-dependent plasticity and time-dependent creep behavior of material. $f$ and $g$ are both neural operators learned from experimental/synthetic data.
